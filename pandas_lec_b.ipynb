## Manipulate columns in Dataframes
### Rename columns in a DataFrame
Example: Rename the column Survived to survived and the Column PassengerId to id.

```python
df7 = df6.iloc[0:100]
print(df7.head())
df7.rename(columns={
    'Survived': 'survived',
    'PassengerId': "id"
}, inplace=True) # Default is False. With this dosen't need to assign it to a new variable
print(df7.head())
```

### **Create a new column**
*Structure*: `df['name_of_new_column'] = new_data`
1. Be careful, you must have the same length in your new data and dataframe otherwise you will get an error.
2. If you want to store the same value just define it in the new column and all the rows of the DataFrame will be assigned with that value.

Example: We want to create an new column with name survived_bool where have True and False.


```python
df7['survived_bool'] = df7['survived'].astype(bool)
df7[['survived', 'survived_bool']]
```

### Drop columns in a DataFrame
*Simple Structure*: `df = df.drop(columns= 'column_name')`
*Structure for multiple columns*: `df = df.drop(columns= ['column_name1', 'column_name2'])`

Example: delete the column survived_bool and the column Cabin.

```python
df7.drop(columns=['Cabin', 'survived_bool'], inplace=True)
df7.head()
```

### Drop rows in a DataFrame
*Simple Structure*: `df = df.drop(index = 5)`

*Structure for multiple columns*: `df = df.drop(index = range(50, 100)])`

Example: Delete the rows from 50 to 100.

```python
df7 = df7.drop(index = range(50, 100))
df7.tail()
```

## **Filtering and Conditional Selection**
### Filter rows based on condition.
*Simple Structure*: `value_name = df[df['column_name'] == 'specific_value']`

*Structure for multiple columns*: `value_name = df[(df['column_name1'] == 'specific_value1') & (df['column_name2'] == 'specific_value2')]`

Example: Find who survived AND males AND bigger than 18.

```python
df_survived_males = df6[(df6['Survived'] == 1) & (df6['Sex'] == 'male') & (df6['Age'] >= 18)]
print(f"Total adult mens that survived: {len(df_survived_males)} wich is {len(df_survived_males)*100/len(df6):.1f}% of the passengers.")
df_males = df6[(df6['Sex'] == 'male') & (df6['Age'] >= 18)]
print(f"Total adult mens: {df_males}")
print(f"The percentage of adult males who survived {len(df_survived_males)*100/len(df_males):.1f}%")
```

Example 2: Find who survived AND females AND bigger than 18.

```python
df_survived_adult_females = df6[(df6['Survived'] == 1) & (df6['Sex'] == 'female') & (df6['Age'] >= 18)]
print(f"Total adult females that survived: {len(df_survived_adult_females)} wich is {len(df_survived_adult_females)*100/len(df6):.1f}% of the passengers.")
df_females = df6[(df6['Sex'] == 'female') & (df6['Age'] >= 18)]
print(f"Total adult females: {len(df_females)}")
print(f"The percentage of adult females who survived {len(df_survived_adult_females)*100/len(df_females):.1f}%")
```

Example 3: Find the percentage of minors passengers that survived.

```python
df_survived_minors = df6[(df6['Survived'] == 1) & (df6['Age'] < 18)]
df_minors = df6[(df6['Age'] < 18)]
print(f"Total minors passengers: {len(df_minors)}")
print(f"The percentage of minors who survived {len(df_survived_minors)*100/len(df_minors):.1f}%")
```

### Filter using **.isin**():

Example: Find the orders from the cities New York, San Francisko, Los Angeles. 

```python
import numpy as np
pc_shop = np.array([
    [1001, 'Laptop', 1, 1200, 'New York'],
    [1002, 'Headphones', 2, 100, 'Los Angeles'],
    [1003, 'Monitor', 1, 300, 'Chicago'],
    [1004, 'Laptop', 1, 1200, 'Houston'],
    [1005, 'Mouse', 3, 25, 'Phoenix'],
    [1006, 'Keyboard', 2, 50, 'Dallas'],
    [1007, 'Monitor', 1, 300, 'San Antonio'],
    [1008, 'Laptop', 2, 1200, 'Austin'],
    [1009, 'Headphones', 1, 100, 'Jacksonville'],
    [1010, 'Mouse', 4, 25, 'San Francisco']
])

df_pc = pd.DataFrame(pc_shop, columns=['orderId', 'product', 'Quantity', 'Price', 'city']) 
print(df_pc)
# Filter with .isin
target_cities = ['New York', 'San Francisco', 'Los Angeles']
city_orders = df_pc[df_pc['city'].isin(target_cities)]
print(f"\n\nOrders from Target Cities (New York, San Francisco, Los Angeles) :\n {city_orders}")
```

### Using `~`() for Negation:

Example: Show all the products except Laptop. 

```python
non_laptops = df_pc[~(df_pc['product']=='Laptop')]
print(f"Orders without Laptops:\n {non_laptops}")
```

Example: how="any"

```python
print(df)
dfdf1= df.dropna(how='any')
print("\n", dfdf1)
print(f"\nThe final lenght of the cleaned dataset is: {len(dfdf1)}")
# For axis= 1, columns
dfdf2= df.dropna(how='any', axis=1)
print("\n", dfdf2)
print(f"\nThe final lenght of the cleaned dataset is: {len(dfdf2.columns)}") 
```

Example with threshold:

```python
print(df)
# Να ελεξει αν τουλάχιστον 3 γραμμες έχουν non null για να το κρατίσει.
dfdf1= df.dropna(thresh=3)
print("\n", dfdf1)
print(f"\nThe final lenght of the cleaned dataset is: {len(dfdf1)}")
# Να ελεξει αν τουλάχιστον 4 γραμμες έχουν non null αλλιώς έξω απο το dataset.
dfdf1= df.dropna(thresh=4)
print("\n", dfdf1)
print(f"\nThe final lenght of the cleaned dataset is: {len(dfdf1)}")
```

Example with threshold with percentage:

```python
print(df)
# Να κρατίσει τις γραμμες που τουλάχιστον το 75% είναι non null.
dfdf1= df.dropna(thresh=0.75*len(df))
print("\n", dfdf1)
print(f"\nThe final lenght of the cleaned dataset is: {len(dfdf1)}")
# Να κρατίσει τις γραμμές που μονο το 100% των στοιχίων τους δεν έχουν not null.
dfdf1= df.dropna(thresh=len(df.columns))
print("\n", dfdf1)
print(f"\nThe final lenght of the cleaned dataset is: {len(dfdf1)}")
```

### **fillna**(): Fill missing values with a specified value or method

<table>
  <thead>
    <tr style="background-color: #E69536; color: white;">
      <th>Parameter</th>
      <th>Description</th>
      <th>Example</th>
    </tr>
  </thead>
  <tbody>
    <tr>
      <td><b>value</b></td>
      <td>Scalar, dict, or Series to use for filling missing values</td>
      <td><code>df.dropna(axis=1)</code></td>
    </tr>
    <tr>
      <td><b>method</b></td>
      <td>Fill using method: 'ffill' (forward fill) or 'bfill' (backward fill) </td>
      <td><code>df.fillna(method='ffill')</code></td>
    </tr>
    <tr>
      <td><b>axis</b></td>
      <td>Fill across rows (0, default-column wise) or columns (1-row wised). <b>There is a difference between dropna and fillna</b></td>
      <td><code>df.fillna(method='bfill', axis=0)</code></td>
    </tr>
    <tr>
      <td><b>limit</b></td>
      <td>Maximum number of consecutive NaNs to fill</td>
      <td><code>df.fillna(method='ffill', limit=2, axis=1)</code></td>
    </tr>
    <tr>
      <td><b>inplace</b></td>
      <td>If True, modifies the DataFrame directly; if False, returns a new object</td>
      <td><code>df.fillna(value='None', inplace=True)</code></td>
    </tr>
  </tbody>
</table>


Example: 

```python
dfdf1 = df.copy()
print(dfdf1)
# Να κρατίσει τις γραμμες που τουλάχιστον το 75% είναι non null.
dfdf1['ages']= df['ages'].fillna(value=int(df['ages'].mean()))
print("\n", dfdf1)
```
