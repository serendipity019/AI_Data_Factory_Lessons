```python
import pandas as pd

# Define the filename
filename = r'../../../AIDataFactory/pdf-files/Python/datasets/lecture_7_dataset/titanic_dataset.csv'
# Read the file
df_titanic = pd.read_csv(filename)
#print the first 5 rows
df_titanic.head()
```

Renaming of collumns:

```python
titanic_cols = list(df_titanic.columns) 
new_titanic_cols = titanic_cols.copy()
new_titanic_cols[6] = 'siblings_spouses_aboard'
new_titanic_cols[7]= 'parents_children'
print(new_titanic_cols)
new_titanic_cols = [name[:1].lower() + name[1:] for name in new_titanic_cols]
print(new_titanic_cols)
for i, name in enumerate(new_titanic_cols):
    df_titanic = df_titanic.rename(columns= {df_titanic.columns[i] : name})

print(df_titanic.columns)
# df_titanic.columns = new_titanic_cols # The easyest all
```

<p style= "text-align: center" > Advanced Filtering</p>

```python

filtering = df_titanic[(df_titanic['survived'] == 0) & (df_titanic['age'] > 30) & ((df_titanic['pclass'] == 1) | (df_titanic['pclass'] == 2))]
filtering.loc[:, ['name', 'age', 'pclass', 'fare', 'survived']]
```

<p style= "text-align: center" > Aditional Filtering</p

```python
third_class_females = df_titanic[(df_titanic['sex'] == 'female') & (df_titanic['pclass'] == 3)]
print(f"Third class females passengers: \n {third_class_females}")
```

```python
passengers_with_siblings = df_titanic[df_titanic['siblings_spouses_aboard'] >= 1]
print(f"\n\n Pasengers with siblings: \n {passengers_with_siblings}")
```

<p style= "text-align: center" > Handling missing data</p

```python
print(df_titanic['age'].median())
df_titanic['age'] = df_titanic['age'].fillna(value= df_titanic['age'].median())
df_titanic.info()
```

```python
# Without the rows where have NaN in embarking column.
df_titanic = df_titanic.dropna(how="any", subset= "embarked")
df_titanic.info()
```

```python
# remove every column that have more than 50%  of the rows NaN
df_titanic = df_titanic.dropna(axis=1, thresh = 0.5 * len(df_titanic))
df_titanic.info()
```
