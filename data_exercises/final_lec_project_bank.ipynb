```python
import pandas as pd
from typing import Literal

class DataAnalyzer:
    """
    A class used to load, handle, filter and analyze a dataset.

    
    Args: 
        file_path (str): The file path to the CSV data source.
    
    Inner attributes:
        data (pd.DataFrame): The main pandas DataFrame containing the loaded dataset.
        filtered_data (pd.DataFrame): A temporary DataFrame storing filtered results.
    """    

    def __init__(self, file_path: str):
        """
        Initializes the CovidDataAnalyzer with a CSV file path.

        Args:
            file_path (str): The path to the CSV file.
        """
        self.data = pd.DataFrame()
        self.filtered_data = pd.DataFrame()
        self.file_path = file_path

    def load_data(self) -> None:
        """Load data from a CSV file"""
        import pandas as pd

        self.data = pd.read_csv(self.file_path)
        print("The data loaded succesfully.")
        # For debuf
        print(type(self.data))

    # Helping method
    def _check_data(self) -> bool:
        if self.data.empty:
            print("No data provided")
            return False
        return True

    # This needing because I taken TypeError object is not subscritable. 
    def __getattr__(self, name):
        if name == "data":
            raise AttributeError(name)
        return getattr(self.data, name)

    # This needs to can code len(data_name)
    def __len__(self) -> int:
        """
        Return the length of the Dataframe.
        """
        return len(self.data)

    def __setitem__(self, key: str, value) -> None:
        self.data[key] = value

    def __getitem__(self, key) -> pd.DataFrame:
       return self.data[key]

    def _check_column_name(self, column_name: str) -> bool:
        column_names = list(self.data.columns)
        if column_name in column_names:
            return True
        
        print(f"The column name {column_name} dosen't exist.")
        return False
    
    def show_head(self, row_num: int= 10) -> pd.DataFrame | None: 
        """
        Override the pandas head. Return the firsts rows of the Dataset.

        Args:
            row_num (int): Number of the first rows. Default = 10
        """
        if not self._check_data():
            return

        print(f"\nThe first {row_num} rows of the DataFrame:\n")
        return self.data.head(min(row_num, len(self.data)))

    
    def show_shape(self) -> None:    
        """
        Override the pandas shape. Print the dimension of the Dataset. 
        """
        if not self._check_data():
            return

        print(f"The DataFrame have {self.data.shape[0]} rows and {self.data.shape[1]} columns.\n")

    def show_sample(self, row_num: int =10) -> pd.DataFrame | None:
        """
        Override the pandas sample. Return random rows of the Dataset.

        Args:
            row_num (int): Number of the random rows. Default = 10
        """
        if not self._check_data():
            return

        print(f"\nA sample of {row_num} rows of the DataFrame:\n")
        return self.data.sample(min(row_num, len(self.data)))

    def show_tail(self, row_num: int =10) -> pd.DataFrame | None:
        """
        Override the pandas tail. Return last rows of the Dataset.

        Args:
            row_num (int): Number of the last rows. Default = 10
        """
        if not self._check_data():
            return

        print(f"\nThe last {row_num} rows of the DataFrame:\n")
        return self.data.tail(min(row_num, len(self.data)))

    def show_info(self) -> str:
        """
        Override the pandas info. Return some info of the Dataset.
        """
        if not self._check_data():
            return

        print(f"\nThe names of the columns and some info from the Dataset are:\n")
        return self.data.info()
    
    def total_null(self) -> None:
        """
        Print how many null values has every column of the Dataset.
        """
        if not self._check_data():
            return

        print(f"\nThe total NaN values per column are:\n")
        print(self.data.isna().sum())

    def show_column_values(self, column_name: str) -> None:
        """
        Print the amount of the different values of a specific column.

        Args:
            column_name (str): The column name of the Dataset.
        """
        if not self._check_data():
            return

        if not self._check_column_name(column_name):
            return

        print(f"\nThe Column {column_name} has the amount of the values below:\n")
        print(self.data[column_name].value_counts(dropna=False))

    def show_describe(self) -> pd.DataFrame | None:
        """
        Call the describe from pandas lib.   
        """
        if not self._check_data():
            return
        
        print("\nSome statistics from the numeric values of the Dataset:\n")
        return self.data.describe()

    # Θα κρατήσω και αυτήν ώστε με μία εντολή να καλώ 4 βασικές μεθόδους ώστε να βλέπω στα γρήγορα τα δεδομένα μου. 
    def describe_data(self) -> pd.DataFrame | None:
        """The first view of the Dataset """
        if not self._check_data():
            return

        self.show_shape()
        self.show_info() 
        print(self.show_describe())
        print(self.show_sample()) 

    def capitalize_categorical_data(self, column_name: list[str]) -> None:
        """
        Make the columns with categorical data to have consistency.

        Args:
            column_name (list[str]): The name of columns with categorical data.

        Examples: 
                male -> Male
                MALE -> Male
        """
        if not self._check_data():
            return

        for col in column_name:
            if not self._check_column_name(col):
                return

            self.data[col] = self.data[col].str.strip().str.lower().str.capitalize()
            print(f"The data of the column {col} capitalized succesfully.")

    def rename_column(self, column_name: str, new_col_name: str) -> None:
        """
        Rename a specific column of the dataset.

        Args:
            column_name (str): The column name you want to change name,
            new_col_name (str): The new column name.  
        """
        if not self._check_column_name:
            return

        self.data = self.data.rename(columns= {column_name: new_col_name})
        print(f"The column name {column_name} has changes sucessfully to: {new_col_name}.")

    def rename_columns(self, column_dict: dict[str]) -> None:
            """
            Rename a specific column of the dataset.
    
            Args:
                column_dict (dict[str]): The dictionary with the columns you want to change name
            
            Example: column_dict = {"col_name1" : "new_col1", "col_name2" : "new_col2" }
                  
            """
            if not self._check_column_name:
                return
    
            self.data = self.data.rename(columns= column_dict)
            print("Your columns name has changes sucessfully!")

    def conditionize_data(self, column_name: str, true_value: str, false_value: str) -> None:
        """
        Make the categorical features that have only two values: 0 or 1. O: False, 1: True
        Carefully if your data dosen't have consistency and have more than 2 values then all the values except tag_value will be None.

        Args:
            column_name (str): The column name of the column where you want to proceed. 
            true_value (str): The name of the value where is the True condition -> 1.
            false_value (str): The name of the value where is the False condition -> 0.

        Example:
            Male -> 1
            Female -> 0
            Other or None -> None
        """ 
        if not self._check_data():
            return

        if not self._check_column_name(column_name):
            return

        counts = self.data[column_name].value_counts(dropna=False)
        unique_values = list(counts.index[:])

        # Here I check if the user want to continue with unclear data.
        if len(unique_values) > 2: 
            print("Your data contains more than 2 unique values or Nulls.")
            print(f"You have these unique values: {unique_values}")
            willness = str(input("\n\nYou want to proceed? Yes if you want."))
            if not willness[0].strip().lower() == "y":
                return

        if not true_value in unique_values:
            print(f"Your True value with name {true_value} dosen't exist in the column values.")
            return
        
        # Edo prospatho na kano enan atypo katharismo gia na sothei isos kapoia amelis katastasi.
        lt_v = str(true_value).strip().lower()
        lf_v = str(false_value).strip().lower()

        def _one_zero(val) -> int | None:
            if pd.isna(val):
                return None
            
            lower_value = str(val).strip().lower()
            if lower_value == lt_v:
                return 1
            elif lower_value == lf_v:
                return 0            
            return None

        self.data[column_name] = self.data[column_name].apply(_one_zero)   
        print("Your data condincionized succesfully!")
        self.show_column_values(column_name)    

    def show_specific_rows_columns(self, column_name: list[str], start_row: int | None = None,
                                    end_row: int | None = None ) -> pd.DataFrame | None:
        """
        This used to appear specific parts of the DataSet.

        Args: 
            column_name (list[str]): A list with the column names that want to appears.
            start_row (int): The integer number of the row where want to start the appearence of the data. Default = None
            end_row (int): The integer number of the row where want to end the appearence of the data. Default = None
        """
        if not self._check_data():
            return
        
        for col in column_name:
            if not self._check_column_name(col):
                return

        if isinstance([start_row, end_row], int) :
            if end_row > len(self.data):
                end_row = len(self.data)
            self.filtered_data = self.data.loc[ start_row : end_row, column_name]
        elif isinstance(start_row, int) and not isinstance(end_row, int):
            self.filtered_data = self.data.loc[ start_row : , column_name]
        else:
            self.filtered_data = self.data.loc[  : , column_name]

        return self.filtered_data


    def handle_missing_values(self, column_name: str) -> None :
        """
            Filling missing values base the type of the data.        
            Args:
                column_name (str): The name of column that contain missing values where want to handle. 
        """
        column_data = self.data[f"{column_name}"].dtype

        if  pd.api.types.is_numeric_dtype(column_data):
            self.data[f"{column_name}"] = self.data[f"{column_name}"].fillna(0)
            print(f"The NaN values in the column {column_name} have been set to 0.")

        elif pd.api.types.is_string_dtype(column_data):
            self.data[f"{column_name}"] = self.data[f"{column_name}"].fillna("Unknown")
            print(f"The NaN values in the column {column_name} have been set to 'Unknown'.")

        else:
            "For boolean and date type"
            self.data[f"{column_name}"] = self.data[f"{column_name}"].fillna(None)
            print(f"The NaN values in the column {column_name} have been set to None.")

    def filter_by_columns_value(self, column_name_value: dict[str],
                                f_case: Literal["==", "!=", "<=", ">=", "<", ">"] , f_option: Literal["&", "|"] = "&") -> pd.DataFrame | None:
        """
        Filter DataFrame columns by specific values. Four columns maximum filter.

        Args:
            column_name_value (dict[str]): The column name and the specific value that you want to filter.
            f_case (Literal["==", "!=", "<=", ">=", "<", ">"]): The symbol of what case to search. 
            f_option (Literal["&", "|"]): For AND use & , for OR use | .
        """
        if not self._check_data():
            return
        
        for col in column_name_value.keys():
            if not self._check_column_name(col):
                return

        length = len(column_name_value)

        match length:
            case 1:
                self.filtered_data = self.data[self.data[column_name_value.keys()] == column_name_value.values()]
            case 2:
                pass    
            case 3:
                pass
            case 4:
                pass
            case _:
                print("Make custom filter, this filter is until 4 cases.")  

       
    def save_filtered_data(self, filename: str) -> None:
        """
        Save in csv file the temporary filtered data.
        
        Args:
            filename: The file name of the filtered data.
        """
        self.filtered_data.to_csv(f"{filename}.csv", index=False)
        print(f"The temporery filtered data saved persistant in the file with name: {filename}.csv .")

    def save_data(self, filename: str) -> None:
        """
        Save in csv file the cleared data.
        
        Args:
            filename: The file name of the cleared data.
        """
        self.data.to_csv(f"{filename}.csv", index=False)
        print(f"The cleared data saved persistant in the file with name: {filename}.csv .")

    
```
