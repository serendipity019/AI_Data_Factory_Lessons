```python
class DataAnalyzer:
    """
    A class used to load, handle, filter and analyze a dataset.

    
    Args: 
        file_path (str): The file path to the CSV data source.
    
    Inner attributes:
        data (pd.DataFrame): The main pandas DataFrame containing the loaded dataset.
        filtered_data (pd.DataFrame): A temporary DataFrame storing filtered results.
    """    

    def __init__(self, file_path: str):
        """
        Initializes the CovidDataAnalyzer with a CSV file path.

        Args:
            file_path (str): The path to the CSV file.
        """
        import pandas as pd

        self.data = pd.DataFrame()
        self.filtered_data = pd.DataFrame()
        self.file_path = file_path

    def load_data(self) -> None:
        """Load data from a CSV file"""
        import pandas as pd 

        self.data = pd.read_csv(self.file_path)
        print("The data loaded succesfully.")
        # For debuf
        print(type(self.data))
    
    def _check_data(self) -> bool:
        if self.data.empty:
            print("No data provided")
            return False
        return True
    
    def show_head(self, row_num: int= 10) -> None: 
        """
        Override the pandas head. Print the firsts rows of the Dataset.

        Args:
            row_num (int): Number of the first rows. Default = 10
        """
        if not self._check_data():
            return

        print(f"\nThe first {row_num} rows of the DataFrame:\n")
        print(self.data.head(min(row_num, len(self.data))))

    
    def show_shape(self) -> None:    
        """
        Override the pandas shape. Print the dimension of the Dataset. 
        """
        if not self._check_data():
            return

        print(f"The DataFrame have {self.data.shape[0]} rows and {self.data.shape[1]} columns.\n")

    def show_sample(self, row_num: int =10) -> None:
        """
        Override the pandas sample. Print random rows of the Dataset.

        Args:
            row_num (int): Number of the random rows. Default = 10
        """
        if not self._check_data():
            return

        print(f"\nA sample of {row_num} rows of the DataFrame:\n")
        print(self.data.sample(min(row_num, len(self.data))))

    def show_tail(self, row_num: int =10) -> None:
        """
        Override the pandas tail. Print last rows of the Dataset.

        Args:
            row_num (int): Number of the last rows. Default = 10
        """
        if not self._check_data():
            return

        print(f"\nThe last {row_num} rows of the DataFrame:\n")
        print(self.data.tail(min(row_num, len(self.data))))

    def show_info(self) -> None:
        """
        Override the pandas info. Print some info of the Dataset.
        """
        if not self._check_data():
            return

        print(f"\nThe names of the columns and some info from the Dataset are:\n")
        self.data.info()
    
    def total_null(self) -> None:
        """
        Print how many null values has every column of the Dataset.
        """
        if not self._check_data():
            return

        print(f"\nThe total NaN values per column are:\n")
        print(self.data.isna().sum())

    def column_values(self, column_name: str) -> None:
            """
            Print the amount of the different values of a specific column.

            Args:
                column_name (str): The column name of the Dataset.
            """
            print(f"\nThe Column {column_name} has the amount of the values below:\n")
            print(self.data[column_name].value_counts())

        

    def describe_data(self) -> None:
        """The first view of the Dataset """
        if not self._check_data():
            return

        self.show_shape()
        self.show_sample() 
        self.show_info() 
        
        print(f"\nSome statistics from the numeric values of the Dataset:\n {self.data.describe()}")
    
    def handle_missing_values(self, column_name: str) -> None :
        """
            Filling missing values base the type of the data.        
            Args:
                column_name (str): The name of column that contain missing values where want to handle. 
        """
        import pandas as pd
        column_data = self.data[f"{column_name}"].dtype

        if  pd.api.types.is_numeric_dtype(column_data):
            self.data[f"{column_name}"] = self.data[f"{column_name}"].fillna(0)
            print(f"The NaN values in the column {column_name} have been set to 0.")

        elif pd.api.types.is_string_dtype(column_data):
            self.data[f"{column_name}"] = self.data[f"{column_name}"].fillna("Unknown")
            print(f"The NaN values in the column {column_name} have been set to 'Unknown'.")

        else:
            "For boolean and date type"
            self.data[f"{column_name}"] = self.data[f"{column_name}"].fillna(None)
            print(f"The NaN values in the column {column_name} have been set to None.")

       
    def save_filtered_data(self, filename: str) -> None:
        """
        Save in csv file the temporary filtered data.
        
        Args:
            filename: The file name of the filtered data.
        """
        self.filtered_data.to_csv(f"{filename}.csv", index=False)
        print(f"The temporery filtered data saved persistant in the file with name: {filename}.csv .")

    
```
