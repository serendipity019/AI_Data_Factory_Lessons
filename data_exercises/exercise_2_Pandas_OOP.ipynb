```python
class CovidDataAnalyzer:
    """
    A class used to load, handle, filter and analyze global COVID-19 dataset.

    
    Args: 
        file_path (str): The file path to the CSV data source.
    
    Inner attributes:
        data (pd.DataFrame): The main pandas DataFrame containing the loaded dataset.
        filtered_data (pd.DataFrame): A temporary DataFrame storing filtered results.
    """    

    def __init__(self, file_path: str):
        """
        Initializes the CovidDataAnalyzer with a CSV file path.

        Args:
            file_path (str): The path to the CSV file.
        """
        import pandas as pd

        self.data = pd.DataFrame()
        self.filtered_data = pd.DataFrame()
        self.file_path = file_path

    def load_data(self) -> None:
        """Load data from a CSV file"""
        import pandas as pd 

        self.data = pd.read_csv(self.file_path)
        print("The data loaded succesfully.")
        # For debuf
        print(type(self.data))


    def describe_data(self) -> None:
        """The first view of the Covid 19 Dataset """
        # Εδώ θα μου άρεσε να έβαζα και κάποιες υπομεθόδους που να εκαναν τα ίδια πράγματα με την describe,
        # την info κ.ο.κ ωστε να μην δίνει παντα υπερβολική πληροφορία με την μία.
        if len(self.data) == 0:
            print("No data provided")

        print(f"The Covid_19 Data have {self.data.shape[0]} rows and {self.data.shape[1]} columns.\n")

        print("\nThe first 10 rows of the DataFrame:\n")
        print(self.data.head(10))
        print("\n\nA sample of 10 rows of the DataFrame:\n")
        print(self.data.sample(10))
        print("\n\nThe last 10 rows of the DataFrame:\n")
        print(self.data.tail(10))
        
        print(f"\n\nThe names of the columns and some info from the covid 19 Dataset are:\n")
        print(self.data.info())
        print(f"\n\nSome statistics from the numeric values of the Covid 19 Dataset:\n {self.data.describe()}")

    # Δεν χρησιμοποιώ το inplace γιατί μου το βγάζει ώς depricated method και στο documentation αναφέρει ως καλή πρακτική να το δηλώνουμε σε καινούργια ή την ίδια μεταβλητή.
    def handle_missing_values(self, column_name: str) -> None :
        """
            Filling missing values base the type of the data.        
            Args:
                column_name (str): The name of column that contain missing values where want to handle. 
        """
        import pandas as pd
        column_data = self.data[f"{column_name}"].dtype

        if  pd.api.types.is_numeric_dtype(column_data):
            self.data[f"{column_name}"] = self.data[f"{column_name}"].fillna(0)
            print(f"The NaN values in the column {column_name} have been set to 0.")

        elif pd.api.types.is_string_dtype(column_data):
            self.data[f"{column_name}"] = self.data[f"{column_name}"].fillna("Unknown")
            print(f"The NaN values in the column {column_name} have been set to 'Unknown'.")

        else:
            "For boolean and date type"
            self.data[f"{column_name}"] = self.data[f"{column_name}"].fillna(None)
            print(f"The NaN values in the column {column_name} have been set to None.")

    def filter_high_cases(self):
        """This method filter the dataset for confirmed cases greater than 100.000
            and deaths above 5000 for identified Countries."""
        high_cases = self.data[(self.data["Confirmed"] > 100_000) & (self.data["Deaths"] > 5_000)
                      & (self.data["Country/Region"] != "Unknown")]
        print("The data filtered")
    
        self.filtered_data = high_cases
        print("The filtered data saved in the temporary memory.")
    
    
    from datetime import date
    def filter_by_date_range(self, start_date: date, end_date: date ) -> None:
        """
            Filter the data by date between a date range. Date format yyyy-mm-dd

            Args:
                start_date (datetime.date): The start date of the filter.
                end_date (datetime.date): The end date of the filter.

            Raise: TypeError if the data has not in the form yyyy-mm-dd
        """
        import pandas as pd 
        self.data['Date'] = pd.to_datetime(self.data['Date'])
    
        if not isinstance(start_date, date) or not isinstance(end_date, date):
            raise TypeError("start_date and end_date must be datetime.date form (yyyy-mm-dd).")
    
        date_range_filter = (self.data['Date']>= pd.Timestamp(start_date )) & (self.data['Date']<= pd.Timestamp(end_date))
        #date_range_filter = (self.data['Date']>= start_date ) & (self.data['Date']<= end_date)
        print("The data filtered")
    
        self.filtered_data = self.data[date_range_filter]
        print("The filtered data saved in the temporary memory.")

    # Custom global statistics 
    # def calculate_global_statistics(self) -> None:
    #     """
    #     Calculates the global total of Confirmed, Deathes and Recovered cases and print the results. 
    #     """
    #     total_global_confirmed_cases = self.data['Confirmed'].sum()
    #     total_global_deaths = self.data['Deaths'].sum()
    #     tot_glob_recov_cases = self.data['Recovered'].sum()
    #     print(f"Total global confirmed cases of covid 19: {total_global_confirmed_cases}\n")
    #     print(f"Total global deaths from covid 19: {total_global_deaths}\n")
    #     print(f"Total global recovered cases from covid 19: {tot_glob_recov_cases}\n")

    # Custom statistics with numpy
    def calculate_np_global_statistics(self) -> None:
        """
        Calculates the global total of Confirmed, Deathes and Recovered cases and print the results. 
        """
        import numpy as np
        total_global_confirmed_cases = np.array(self.data['Confirmed'])
        total_global_deaths = np.array(self.data['Deaths'])
        tot_glob_recov_cases = np.array(self.data['Recovered'])

        print(f"Total global confirmed cases of covid 19: {total_global_confirmed_cases.sum()}\n")
        print(f"Total global deaths from covid 19: {total_global_deaths.sum()}\n")
        print(f"Total global recovered cases from covid 19: {tot_glob_recov_cases.sum()}\n")
    
    def save_filtered_data(self, filename: str) -> None:
        """
        Save in csv file the temporary filtered data.
        
        Args:
            filename: The file name of the filtered data.
        """
        self.filtered_data.to_csv(f"{filename}.csv", index=False)
        print(f"The temporery filtered data saved persistant in the file with name: {filename}.csv .")

    
```
