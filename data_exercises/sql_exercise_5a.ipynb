### Connection with the Database.

```python
import mysql.connector as sql

my_conn = sql.connect(
    host = 'localhost', 
    user = 'root',
    password = MY_SQL_PASSWORD,
    database = 'sakila'
)

if my_conn.is_connected():
    print('Connect Succesfully') 
```

### Υλοποίηση ερωτήματος και φόρτωση του Dataframe.
**Ερώτημα:** Να φέρουμε για κάθε χρήστη το επώνυμο, όνομα και το πλήθος των ενοικιάσεων που έχει κάνει.

```python
query_1 = """ 
select c.last_name as Lastname, c.first_name as Firstname, count(r.rental_id) as "Total Rents"
from customer as c inner join rental as r on r.customer_id = c.customer_id
group by Lastname, Firstname ; 
"""
```

```python
import pandas as pd

df = pd.read_sql(query_1, my_conn)
print(df)
```

### Έκανα στον πίνακα Language το γνώρισμα last_update μη υποχρεωτικό μέσα απο το workbench για ευκολία.
Παρακάτω θα εισάγω τρείς ακόμα γλώσσες.

```python
import_query = """
insert into language (language_id, name) values
(7, "Spanish"),
(8, "Greek"),
(9, "Arabic")
"""
```

```python
import_cursor = my_conn.cursor()
import_cursor.execute(import_query)
print('Done! The elements imported successfully.')
my_conn.commit() # χωρίς αυτό δεν φαίνονται οι αλλαγές στο workbench.
import_cursor.close()
```

Παρακάτω βλέπουμε τα κενούργια δεδομένα στον πίνακα language.

```python
language_cursor = my_conn.cursor()
language_cursor.execute('select * from language;')
language = language_cursor.fetchall()

for row in language:
    print(row)

language_cursor.close()
```

Με το παρακάτω skript-ακι διαγράφω ότι καινούργιο μπήκε στον πίνακα language (Αρχικά είχε μόνο 6 γλώσσες.)

```python
delete_cursor = my_conn.cursor()
delete_cursor.execute('delete from language where language_id > 6')
print("Done")
my_conn.commit()
delete_cursor.close()
```
