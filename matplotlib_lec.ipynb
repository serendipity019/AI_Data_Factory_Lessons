<h3 style = "text-align: center"> <b>Creation of a scatterplot</b> </h3>
<p><b>What is Scatter Plot:</b>
<ul><li>A scatter plot shows the relationship between two variables as points on a 2D grid.</li>
    <li>Useful for identifying trends, clusters and outliers.</li>
</ul>
</p>
<p>Example:</p>

```python
# Dummy data
x = [1, 2, 3, 4, 5]
y = [2, 3, 5, 7, 9]

# Create a scatter plot
plt.scatter(x,y, color="blue")
plt.title("Dummy Scatter Plot")
plt.xlabel("Variable 1")
plt.ylabel("Variable 2")
plt.show()
```

<h3 style = "text-align: center"> <b>Creation of a lineplot</b> </h3>
<p><b>What is lineplot:</b>
<ul><li>A line plot connects the data points in a sequence, often used for showing <strong>trends over time</strong>.</li>
</ul>
</p>
<p>Example:</p>

```python
# Dummy data
month = ['January', 'February', 'March', 'April', 'May', 'June']
sales = [150, 180, 220, 300, 350, 400]

# Create a line plot
plt.plot(month, sales, marker='o', linestyle='-', color='green')
plt.title("Dummy Line Plot")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.show()
```

<h3 style = "text-align: center"> <b>Creation of a barplot</b> </h3>
<p><b>What is Bar Plot:</b>
<ul><li>A bar plot displays <strong>categorical data</strong> with rectangular bars representing values.</li>
    <li>Useful for comparing quantities across categories.</li>
</ul>
</p>
<p>Example:</p>

```python
# Dummy data
categories = ['A', 'B', 'C', 'D']
values = [150, 300, 220, 340]

# Create a bar plot
plt.bar(categories, values, color='skyblue')
plt.title("Dummy Bar Plot")
plt.xlabel("Categories")
plt.ylabel("Values")
plt.show()
```

<h3 style = "text-align: center"> <b>Creation of a histogram</b> </h3>
<p><b>What is Histogram:</b>
<ul><li>A histogram displays the<strong>distribution</strong>of a numerical variable by grouping values into bins.</li>
    <li>Useful for understanding the shape and spread of data.</li>
</ul>
</p>
<p>Example:</p>

```python
# Dummy data
import numpy as np
data = np.random.normal(50,15,100)

# Create a line plot
plt.hist(data, bins=10, color='purple', edgecolor='black')
plt.title("Dummy Histogram")
plt.xlabel("Value")
plt.ylabel("frequency")
plt.show()
```

### Small Exercise 2:
- Load the Titanic dataset
- Create a bar plot showing the number of passengers in each passenger class (Pclass).
- Label the x-axis as "Passenger Class" and y-axis as "Number of Passengers".
- Create a proper title for that plot.

```python
import pandas as pd

# Define the filename
filename = r'../../AIDataFactory/pdf-files/Python/datasets/lecture_7_dataset/titanic_dataset.csv'
# Read the file
df_titanic = pd.read_csv(filename)
#print the first 5 rows
df_titanic.head()
```

```python
pclass_counts = df_titanic['Pclass'].value_counts().sort_index()

categories = pclass_counts.index
values = pclass_counts.values

plt.bar( categories, values)
plt.title("Passengers per Class")
plt.xlabel("Passenger Class")
plt.ylabel("Number of Passengers")
plt.gca().set_xticks([1,2,3])
plt.show()
```

## Advanced plotting with Matplotlib – Object Oriented Interface

### What Are fig and ax?
- **fig:** The **Figure** object — the entire canvas or window that can contain one or more plots. 
- **ax:** The **Axes** object — the actual **plot area** where data is drawn (including axis, labels, ticks, etc.).

**Hint**: Think of fig as the **sheet of paper**, and ax as a **subplot drawn on that paper**.

**Why Use This Interface?**
- Better for **complex layouts**, subplots, and reusable plotting functions.
- Cleaner control over titles, axis labels, legends, ticks, etc.

# Small Exercise 3:
- Load The Titanic dataset.
- Create a scatter plot of Age vs Fare.
- Label the x-axis as "Age" and y-axis as "Fare".
- Give a proper title for that. 
- (optional) Save the plot in a .pdf file

**Note**: Implementation in Jyputer is suggested.

```python
import pandas as pd

# Define the filename
filename = r'../../AIDataFactory/pdf-files/Python/datasets/lecture_7_dataset/titanic_dataset.csv'
# Read the file
df_titanic = pd.read_csv(filename)
#print the first 5 rows
df_titanic.head()
```

```python
age_fare = df_titanic.dropna(subset=['Age', 'Fare'])
age_fare.reset_index()
age_fare = age_fare.loc[: ,['Age', 'Fare']]
```

```python
x = age_fare['Age']
y = age_fare['Fare']

plt.scatter(x, y)
plt.title('Age to Fare Scatter Plot')
plt.xlabel('Age')
plt.ylabel('Fare')
#plt.savefig('AgeVsFare_Scatter.pdf')
plt.show()
```
