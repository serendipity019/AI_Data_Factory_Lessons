## Data Structures:
Pandas has two main data structures it uses: **Series** and **DataFrames**
### **Series** (`pd.Series()`)
- It is one-dimensional labeled array, similar to a list or column in Excel.
- It consists of data values and an associated index (labels), making it easy to reference values by their index

Example 1: Create a Series from a list.

```python
# define a list
my_list = [10, 20, 2, 3, 17, 23, 0]

# Create a series
df_series = pd.Series(my_list)
print(df_series)
```

Example 2: Create a Series from a dictionary.

```python
# Creating a dictionary
data_dict = {'a': 10, 'b': 20, 'c': 30}
# Create series
series_from_dict = pd.Series(data_dict)
print(series_from_dict)
```

Example 3: Find a value by each position/index. Use **`.iloc[index]`**/ **`.loc['key']`** respectively.

```python
# iloc(index) with list 
value_position_one = df_series.iloc[1]
print(f"In position 1 the value is: {value_position_one}")
# iloc with index
value_position_two_dict = series_from_dict.iloc[2]
print(f"In position 2 the value in dictionary is: {value_position_two_dict}")
key_value_position_b_dict = series_from_dict.loc['b']
print(f"The key value of b in dictionary is: {key_value_position_b_dict}")
```

## **DataFrames** (`pd.DataFrame(data)`)
- A DataFrame is a two-dimensional, labeled data structure in pandas, similar to a table in a
database or an Excel spreadsheet.
- It consists of rows and columns, where:
    - Each column can contain different data types (e.g., integers, strings, floats).
    - Each row has a unique index, which can be used to access or modify specific entries.
- DataFrames are flexible and make it easy to perform various data manipulation operations.

**Key Points about DataFrames**
- Labeled Axes: Each row and column has labels, making it easy to access data by name.
- Data Alignment: DataFrames automatically align data by labels in operations, making them intuitive to
work with.
- Data Source Compatibility: DataFrames can be created from various data sources like dictionaries,
lists, CSV files, Excel sheets, SQL databases, and more.

Example 1: Create from lists with the help of a dictionary.

```python
# We have 4 lists with data
names = ['Rafail', 'Katerina', 'Giorgos', 'Maria']
ages = [25, 22, 32, None]
cities = ['Voula', 'Zografou', 'Marousi', 'Pagrati']
grades = [92, 86, None, 91]
# Create a dictionary
data = {
    'names': names,
    'ages' : ages,
    'cities': cities,
    'grades': grades
}
# Create dataframe
df = pd.DataFrame(data)
print(df)
```

Example 2: Create dataframe from lists of list.

```python
my_movies = [
    ['Shrek', 'Giorgos', 1997],
    ['Kill Bill 1', 'Tarantino', 1988],
    ['Avengers Endgame', 'Rouso', 2024],
    ['The Titanic', 'Cameron', 2002]
]

df2 = pd.DataFrame(data=my_movies, columns=['name', 'author', 'year']) 
print(df2)
```

Example 3:Create dataframe from a dictionary.

```python
shop_data = {
    'Product': ['Laptop', 'Tablet', 'Smartphone'],
    'Price' : [1200, 300, 800],
    'Stock': [25, 100, 50]
}

df3 = pd.DataFrame(shop_data)
print(df3)
```

Example 4: Create dataframe from a list of Dictionaries

```python
sales_data = [
    {'Month': 'Jan', 'Sales': 1500, 'Profit': 300},
    {'Month': 'Feb', 'Sales': 1700, 'Profit': 400},
    {'Month': 'Mar', 'Sales': 1600, 'Profit': 350},
]
df4 = pd.DataFrame(sales_data)
print(df4)
```

Example 5 (Very useful): Create Dataframe from NumPy array.

```python
import numpy as np

arr = np.array([
    [25.3, 60.2, 5.1],
    [26.1, 55.8, 4.9],
    [23.7, 63.5, 5.5]
])

df5= pd.DataFrame(arr, columns=['Temperature (c)', 'Humidity (%)', 'Wind Speed (m/s)'])
print(df5)
```

**.loc [Label-Based Selection]**
- **Purpose**: Selects data by label. It’s primarily used when you know the row or column labels you
want to access.
- **Syntax**: `df.loc[rows, columns]`, `df.loc[:, [columns]]` (for all the samples)
- **Key Points**:
    - You specify row and column labels.
    - It can be used with conditions for filtering.

Example 1: Selecting a single row by label.

```python
df6.loc[0]
```

Example 2: Selecting multiple rows by index.

```python
df6.loc[0:4] # Gets rows from index 0 to 4 (both inclusive)
```

Example 3: Selecting multiple rows by label we desire.

```python
df6.loc[0:4, ['Name', 'Age', 'Survived']]
```

Example 4: Selecting rows by giving a condition ("Survived == 1") And find the percentage of the survivors.

```python
survived_passengers = df6.loc[df6['Survived'] == 1]
print(survived_passengers.head())
print(f"The percentage of survived passenger is: {len(survived_passengers)*100/len(df6):.1f}%")
```

Example 5: We want to see all the lines of the specific columns Age and Survived.

```python
print(df6[["Age", "Survived"]])
print(df6[df6['Survived']==1]) # Make the same with the loc above.
```

**.iloc[Index-Based Selection]**
- **Purpose:** Selects data by position (integer index). It’s useful when you want to retrieve data based
on the numerical position of rows and columns.
- **Syntax:** df.iloc[rows, columns]
- **Key Points:**
    - You specify the positions of rows and columns as integers.
    - Slicing works like Python lists, where the start is inclusive, but the end is exclusive.
    - Commonly used for selecting the first n rows or columns.

Example 1: Selecting a single row by index. Get the first row in the Dataframe.

```python
df6.iloc[0]
```

Example 2: Gets rows from index 20 to 27

```python
df6.iloc[20:28]
```

Example 3: Gets the rows 30 to 40 for the 3 first columns. 

```python
df6.iloc[30:41, 1:4]
```
